```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(bc.data, bc.target, test_size=0.2, random_state=42)
x_train.shape, x_test.shape, y_train.shape, y_test.shape
```

```python
from sklearn.preprocessing import MinMaxScaler
scalar=MinMaxScaler(feature_range=(0,1))
X_train=scalar.fit_transform(x_train)
X_test=scalar.transform(x_test)
```

```python
X_train[200]
```

```python
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
model.fit(X_train, y_train)
```

```python
y_pred_train = model.predict(X_train)
y_pred_test = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
def calculate_metrics(y_train, y_test,y_pred_train, y_pred_test):

    acc_train = accuracy_score(y_true=y_train, y_pred=y_pred_train)
    acc_test = accuracy_score(y_true=y_test, y_pred=y_pred_test)

    p_train = precision_score(y_true=y_train, y_pred=y_pred_train)
    p_test = precision_score(y_true=y_test, y_pred=y_pred_test)

    r_train = recall_score(y_true=y_train, y_pred=y_pred_train)
    r_test = recall_score(y_true=y_test, y_pred=y_pred_test)

    f1_train = f1_score(y_true=y_train, y_pred=y_pred_train)
    f1_test = f1_score(y_true=y_test, y_pred=y_pred_test)

    conf_train = confusion_matrix(y_true=y_train, y_pred=y_pred_train)
    conf_test = confusion_matrix(y_true=y_test, y_pred=y_pred_test)

    print(f"accuracy train: {acc_train}, accuracy test: {acc_test}")
    print()
    print(f"precition train: {p_train}, precition test: {p_test}")
    print()
    print(f"recall train: {r_train}, recall test: {r_test}")
    print()
    print(f"f1 train: {f1_train}, f1 test: {f1_test}")
    print()
    print(f"confusion matrix train:\n{conf_train}")
    print()
    print(f"cofusion matrix test:\n{conf_test}")
    print()

    # return acc_train, acc_test, p_train, p_test, r_train, r_test, f1_train, f1_test
```

```python
calculate_metrics(y_train, y_test,y_pred_train, y_pred_test)
```
